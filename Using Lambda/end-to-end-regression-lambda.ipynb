```python
# import dependencies
import csv
from pathlib import Path

from model import LinearRegression
import joblib
```

```python
# Load data
X = []
y = []

with open("data/exam_scores.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        X.append(float(row["hours_studied"]))
        y.append(float(row["exam_score"]))
```

```python
# Train

model = LinearRegression()

# model.fit(X_train, y_train)

# joblib.dump(model, "model.joblib")

model = LinearRegression()
model.fit(X, y)

print(f"Intercept: {model.intercept:.4f}")
print(f"Slope: {model.slope:.4f}")
```

```python
# Save artifact
Path("model").mkdir(exist_ok=True)

joblib.dump(
    model,
    "model/exam_score_model.joblib"
)

print("Model saved to model/exam_score_model.joblib")
```

```python
# Try to predict using the model

model = joblib.load(
    "model/exam_score_model.joblib"
)


hours = 8


prediction = model.predict(
    [hours]
)

print(prediction[0])
```
