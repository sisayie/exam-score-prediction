```python
# Create Session

import boto3
from sagemaker.core.helper.session_helper import Session

boto_session = boto3.Session(region_name="us-east-1")

sagemaker_session = Session(
    boto_session=boto_session
)

print(sagemaker_session.boto_region_name)
```

```python
# Download dataset

import boto3
import pandas as pd

BUCKET = "my-ml-bucket-2108"
DATA_KEY = "data/exam_scores.csv"

s3 = boto3.client("s3")

s3.download_file(
    BUCKET,
    DATA_KEY,
    "exam_scores.csv"
)

df = pd.read_csv("exam_scores.csv")

print(df)
```

```python
# Train Model

from sklearn.linear_model import LinearRegression

X = df[["hours_studied"]].values
y = df["exam_score"].values

model = LinearRegression()

model.fit(X, y)

print("Coefficient:", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
# Save Trained Model

import joblib

joblib.dump(
    model,
    "exam_model.joblib"
)
```

```python
# Try to predict using the model

import joblib

model = joblib.load(
    "exam_model.joblib"
)

hours = 8

prediction = model.predict(
    [[hours]]
)

print(prediction[0])
```

```python
# Upload Model to s3

s3.upload_file(
    "exam_model.joblib",
    BUCKET,
    "model/exam_model.joblib"
)
```
